## 3. Voix des personnages (gratuit)
Envoyez **un extrait de 10 à 20 s par personnage** (une seule voix, sans musique), nommé comme le
personnage : `BAYE.wav`, `MODOU.wav`, `PETIT_MAMADOU.wav`, `TANTIE_AWA.wav`…
Astuce : tirez-les de **vos** vidéos Grok avec `python run.py voix auto ma_video.mp4`.
N'utilisez que des voix dont vous avez les droits.

```
from google.colab import files
os.makedirs("state/voice_refs", exist_ok=True)
for name, data in files.upload().items():
    open(f"state/voice_refs/{name}", "wb").write(data)
!ls state/voice_refs
```
